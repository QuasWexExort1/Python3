```python
import random

with open("f.txt", "w") as f:
    for _ in range(20):
        f.write(f"{random.randint(1, 10)}\n")

with open("f.txt", "r") as f:
    unique_nums = list(dict.fromkeys(f.read().split()))

with open("g.txt", "w") as g:
    g.write("\n".join(unique_nums))

with open("g.txt", "r") as g:
    print(g.read())
```
